Given the system of inequalities below, determine the shape of the feasible region and find the vertices of the feasible region. Give the shape as "triangle", "quadrilateral", or "unbounded". Report your vertices starting with the one which has the smallest $𝑥$
-value. If more than one vertex has the same, smallest $𝑥$
-value, start with the one that has the smallest $𝑦$
-value. Proceed clockwise from the first vertex. Leave any unnecessary answer spaces blank

\begin{align*}
𝑥+𝑦&\le7\\
2𝑥+𝑦&\le10\\
𝑥&\ge0\\
𝑦&\ge0
\end{align*}

4 equations, 2 variables, thus, 4C2=6 intersection points.

```python
# Finding corner points
a = np.array([[1, 1], [2, 1]])
b = np.array([7, 10])
xy_vec = np.linalg.solve(a,b)
x1, x2 = xy_vec[0], xy_vec[1]
c = ([1,0], [1,1])
d = ([0,7])
vec = np.linalg.solve(c,d)
e = ([0,1],[2,1])
f = ([0,10])
vec1 = np.linalg.solve(e,f)
print(xy_vec, vec, vec1)
```

```python
x = np.linspace(-1, 8)
def y1(x):
  return 7-x
def y2(x):
  return 10-2*x


plt.plot(x, y1(x))
plt.plot(x, y2(x))
plt.scatter([5, y2(5)], [0,0], c='grey')
plt.scatter(x1, y1(x1), c='grey')
plt.scatter(0, y1(0))
plt.axhline(y=0, color='r')
plt.axvline(x=0, color='purple')
plt.fill()
```

```python
x, y = symbols('x'), symbols('y')
obj_func = 3*x+2*y
call_func = lambdify([x,y], obj_func)
points = [[5,y2(5)], [0,0], [x1,y1(x1)], [0,y1(0)]]
for i in points:
  x = i[0]
  y = i[1]
  print(f'3*({x}) + 2*({y})=', call_func(x,y))
```

```python
a = np.array([[26, 13], [66, 55]])
b = np.array([975, 3113])
xy_vec = np.linalg.solve(a,b)
x1, x2 = xy_vec[0], xy_vec[1]
c = ([0, 1], [26,13])
d = ([0,975])
vec = np.linalg.solve(c,d)
e = ([1,0], [66,55])
f = ([0, 3113])
vec2 = np.linalg.solve(e,f)
print(xy_vec, vec, vec2)
```
